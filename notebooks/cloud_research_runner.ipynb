```python
from pathlib import Path
import shutil
from google.colab import files

DRIVE_ROOT = Path('/content/drive/MyDrive/UAV Dynamic Routing')
BUNDLE = DRIVE_ROOT / 'uav_cloud_bundle.zip'
PROJECT = Path('/content/uav-dynamic-routing')
DRIVE_ROOT.mkdir(parents=True, exist_ok=True)
if not BUNDLE.exists():
    uploaded = files.upload()
    uploaded_name = next(iter(uploaded))
    shutil.copy2(uploaded_name, BUNDLE)
if PROJECT.exists():
    shutil.rmtree(PROJECT)
shutil.unpack_archive(BUNDLE, PROJECT)
%cd /content/uav-dynamic-routing
```
